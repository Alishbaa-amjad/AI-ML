## What is Natural Language Processing (NLP)?

Natural Language Processing (NLP) is a branch of artificial intelligence that gives computers the ability to understand, interpret, and manipulate human language.

Humans speak and write in highly complex, unstructured ways filled with slang, grammar rules, and context. Computers, on the other hand, only understand structured numbers (1s and 0s). NLP acts as the translator or "bridge" that takes messy human text and turns it into structured data that a machine learning model can process.

**The Real-World Example: Smart Home Voice Assistant** 

Let's see how NLP works in a practical scenario: You speak to a smart speaker and say:

"Hey, can you turn off the living room lights?"

A computer cannot instantly understand what "living room" or "lights" means. It uses an NLP pipeline to break down and understand your command through these steps:

* Tokenization: The computer breaks your sentence down into individual words or "tokens."

     * ["Hey", "can", "you", "turn", "off", "the", "living", "room", "lights"]

* Stop Words Removal: It filters out common filler words that don't add unique meaning (like "Hey", "can", "you", "the").

     * Remaining crucial words: ["turn", "off", "living", "room", "lights"]

* Part-of-Speech (POS) Tagging: It identifies the grammatical role of each word to understand the action.

     * Turn off = Action (Verb)
     * Living room = Location (Noun)
     * Lights = Object (Noun)

* Named Entity Recognition (NER) / Intent Extraction: The system maps these tagged words to a specific backend command.

     * Intent: SWITCH_POWER_OFF
     * Target Object: light_fixture
     * Target Location: room_zone_04 (Living Room)

## Python Coding Implementation (Using NLTK)

Here is how you write a basic Python script using the NLTK (Natural Language Toolkit) library discussed in your article to perform the first step (Tokenization) on our example sentence:

---

```python
import nltk

# Fix: Download BOTH required packages for the newer NLTK tokenizer
nltk.download('punkt')
nltk.download('punkt_tab')

# Now run your tokenizer smoothly
voice_command = "Hey, can you turn off the living room lights?"
tokens = nltk.word_tokenize(voice_command)

print("--- Raw Input String ---")
print(voice_command)
print("\n--- Processed NLP Tokens ---")
print(tokens)
```

```python
import matplotlib.pyplot as plt
import networkx as nx

# 1. Initialize an empty directed graph
G = nx.DiGraph()

# 2. Define the stages of our NLP pipeline (Nodes)
stages = [
    "Raw Text\nInput",
    "Tokenization",
    "Stop Words\nRemoval",
    "POS Tagging",
    "Intent\nExtraction"
]

# 3. Add connections (Edges) to create the flow
edges = [
    (stages[0], stages[1]),
    (stages[1], stages[2]),
    (stages[2], stages[3]),
    (stages[3], stages[4])
]
G.add_edges_from(edges)

# 4. Set up the layout and visual configurations
# Multi-line positions for a straight top-to-bottom or left-to-right flow
pos = {
    stages[0]: (0, 5),
    stages[1]: (0, 4),
    stages[2]: (0, 3),
    stages[3]: (0, 2),
    stages[4]: (0, 1)
}

plt.figure(figsize=(6, 8))
plt.title("Natural Language Processing (NLP) Pipeline Flow", fontsize=14, fontweight='bold', pad=20)

# 5. Draw the graph components
nx.draw_networkx_nodes(G, pos, node_size=3000, node_color='skyblue', node_shape='s')
nx.draw_networkx_edges(G, pos, edgelist=edges, arrowstyle='->', arrowsize=20, edge_color='gray', width=2)
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold', font_family='sans-serif')

# Remove axis lines for a clean look
plt.axis('off')
plt.tight_layout()

# 6. Display the graph
plt.show()
```
